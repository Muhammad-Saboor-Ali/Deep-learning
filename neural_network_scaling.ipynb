```python
import pandas as pd
import numpy as np
import random
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential
```

```python
df = pd.DataFrame({"age":np.random.randint(18,40,size=10),
             "salary":np.random.randint(25000,50000,size=10)})
```

```python
df["purchased"] = df["salary"].apply(lambda x: 1 if x > 35000 else 0)
```

```python
df
```

```python
x = df.drop('purchased', axis=1)
y = df['purchased']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
x_trained_scale = ss.fit_transform(x_train)
```

```python
x_test_scaled = ss.transform(x_test)
```

```python
model = Sequential()
```

```python
model = Sequential([
    Dense(6, activation='relu', input_shape=(2,)),  # hidden layer
    Dense(1, activation='sigmoid')  # output layer (binary classification)
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.fit(x_trained_scale,y_train, epochs=50)
```
